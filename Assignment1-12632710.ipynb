```python
# 3. Pascal triangle

def Pascal_triangle(k):
    row = [1]

    for i in range(1, k):
        new_row = [1]

        for j in range(len(row) - 1):
            new_row.append(row[j] + row[j + 1])

        new_row.append(1)
        row = new_row

    print(row)


Pascal_triangle(100)
Pascal_triangle(200)
```

```python
# 4. Add or double
import random
#定义函数，输入目标金额 x，初始操作次数为 0，
#只要 x 大于 1，就继续计算，判断 x 是否为偶数，如果是偶数，就除以 2，奇数...
def Least_moves(x):
    moves = 0
    while x > 1:
        if x % 2 == 0:
            x = x // 2
        else:
            x = x - 1
        moves = moves + 1
    print(moves)
Least_moves(2)
Least_moves(5)

for i in range(5):
    x = random.randint(1, 100)
    print("x =", x)
    Least_moves(x)
```

```python
#5.1 Dynamic programming
import itertools
import random

def Find_expression(x):
    digits = "123456789"
    count = 0

    for ops in itertools.product(["+", "-", ""], repeat=8):
        expression = digits[0]

        for i in range(8):
            expression = expression + ops[i] + digits[i + 1]

        if eval(expression) == x:
            print(expression, "=", x)
            count = count + 1

    return count


x = random.randint(1, 100)
print("Target =", x)
print("Total solutions =", Find_expression(x))

```

```python
#5.2 Dynamic programming
import matplotlib.pyplot as plt

Total_solutions = []

for x in range(1, 101):
    count = 0

    for ops in itertools.product(["+", "-", ""], repeat=8):
        expression = "1"

        for i in range(8):
            expression = expression + ops[i] + str(i + 2)

        if eval(expression) == x:
            count = count + 1

    Total_solutions.append(count)

plt.plot(range(1, 101), Total_solutions)
plt.xlabel("Number")
plt.ylabel("Total solutions")
plt.title("Number of solutions for 1 to 100")
plt.show()

maximum = max(Total_solutions)
minimum = min(Total_solutions)

print("Maximum solutions:", maximum)
print("Numbers with maximum solutions:")
for i in range(100):
    if Total_solutions[i] == maximum:
        print(i + 1)

print("Minimum solutions:", minimum)
print("Numbers with minimum solutions:")
for i in range(100):
    if Total_solutions[i] == minimum:
        print(i + 1)
```
